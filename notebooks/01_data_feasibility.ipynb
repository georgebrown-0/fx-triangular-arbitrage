## Is historical data obtainable for free?

Dukascopy.com provides free tick-by-tick data for FX pairs. An example of this data is shown below. Downloads are available for both bid and ask data, but they are separately available. A 'tick' is an update to the price quote, not a trade execution. Data is available for all forex majors and minors, with extensive coverage of liquid exotics. Data generally dates back to 2003-2005. Data is reliably available one hour at a time. The example below only shows second-level resolution, but data is available at lower levels (downloads are slightly more complicated for this).

```python
from datetime import datetime
import dukascopy_python
```

```python
from dukascopy_python import instruments

# Choose a known weekday date (e.g., Wednesday, Jan 15, 2025)
EUR_USD = dukascopy_python.fetch(
    instrument='EUR/USD',
    interval=dukascopy_python.INTERVAL_TICK,
    offer_side=dukascopy_python.OFFER_SIDE_BID,
    start=datetime(2025, 1, 15, 8, 0, 0),
    end=datetime(2025, 1, 15, 9, 0, 0)
)

# Format millisecond timestamps
EUR_USD.head()
```

```python
USD_JPY = dukascopy_python.fetch(
    instrument='USD/JPY',
    interval=dukascopy_python.INTERVAL_TICK,
    offer_side=dukascopy_python.OFFER_SIDE_BID,
    start=datetime(2025, 1, 15, 8, 0, 0),
    end=datetime(2025, 1, 15, 9, 0, 0)
)

EUR_JPY = dukascopy_python.fetch(
    instrument='EUR/JPY',
    interval=dukascopy_python.INTERVAL_TICK,
    offer_side=dukascopy_python.OFFER_SIDE_BID,
    start=datetime(2025, 1, 15, 8, 0, 0),
    end=datetime(2025, 1, 15, 9, 0, 0)
)

EUR_JPY.head()
```

```python
data = {'EUR_USD': EUR_USD, 'USD_JPY': USD_JPY, 'EUR_JPY': EUR_JPY}

for name, d in data.items():
    print(f'name: {name}')
    print(f'length: {len(d)}')
    print(f'first timestamp: {d.index[0]}')
    print(f'last timestamp: {d.index[-1]}')
    print(f'columns names: {d.columns}')
    print(f'columns types: {d.dtypes}')
    print(f'increasing: {d.index.is_monotonic_increasing}')
    print(f'number of duplicate timestamps: {d.index.duplicated().sum()}')
    print(f'number of NaN values: {d.isna().sum()}')
    spread = d['askPrice'] - d['bidPrice']
    print(f'number of negative spread values: {(spread < 0).sum()}')
    print(f'spread min: {spread.min()}')
    print(f'spread median: {spread.median()}')
    print(f'spread max: {spread.max()}')
    print('')
    
```
