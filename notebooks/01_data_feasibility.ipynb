## Is historical data obtainable for free?

Dukascopy.com provides free tick-by-tick data for FX pairs. An example of this data is shown below. Downloads are available for both bid and ask data, but they are separately available. A 'tick' is an update to the price quote, not a trade execution. Data is available for all forex majors and minors, with extensive coverage of liquid exotics. Data generally dates back to 2003-2005. Data is reliably available.
